## Config

```python
import pandas as pd
from pathlib import Path
```

```python
DATA_DIR = Path() / "data" / "raw"

def load_file(file, sep=',', skiprows=0) -> pd.DataFrame:
    df = pd.read_csv(
        DATA_DIR / file, 
        sep=sep,
        low_memory=False,
        encoding='utf-8',
        skiprows=skiprows,
        na_values=['..', 'NA', 'N/A', '']
    ).dropna(how='all')

    show_info(df)
    return df

def show_info(df):
    print("SHAPE:")
    print(df.shape)

    print("\n--------")
    print("INFO:")
    df.info()

    print("\n--------")
    print("SAMPLE:")
    display(df.head())

    print("\n--------")
    print("MISSING VALUES:")
    missing = pd.DataFrame({
        'count': df.isnull().sum(),
        'percentage': (df.isnull().mean() * 100).round(2)
    })
    display(missing[missing['count'] > 0])  # only shows columns WITH missing values

    print("\n--------")
    print("UNIQUE VALUES:")
    print(df.nunique())

    print("\n--------")
    print("STATISTICAL SUMMARY:")
    display(df.describe())
```

## Load data

### Population

```python
df_population = load_file("fao_population_2017.csv")
```

```python
df_population.Value *= 1000
df_population.drop(columns=['Unit', 'Year'], inplace=True)
df_population.rename(columns={ 'Value': 'population' }, inplace=True)
df_population
```

```python
countries_to_remove = [
    'China',                      # generic aggregate
    'China, Hong Kong SAR',
    'China, Macao SAR', 
    'China, Taiwan Province of',
]

df_population = df_population[
    ~df_population['Area'].isin(countries_to_remove)
]
```

### Poultry supply per capita

```python
df_dispo = load_file("fao_disponibility_2017.csv")
```

```python
df_dispo.drop(columns=['Year', 'Item', 'Element'], inplace=True)
df_dispo.rename(columns={ 'Value': 'poultry_supply_capita' }, inplace=True)
```

### Domestic supply

```python
df_domestic = load_file("fao_domestic_supply_2017.csv")
```

```python
df_domestic.Value *= 1000000
df_domestic.drop(columns=['Year', 'Item', 'Element', 'Unit'], inplace=True)
df_domestic.rename(columns={ 'Value': 'poultry_domestic_supply' }, inplace=True)
```

### Chicken import

```python
df_import = load_file("fao_chicken_imports_2017.csv")
```

```python
df_import.Value *= 1000
df_import.drop(columns=['Year', 'Unit'], inplace=True)
df_import.rename(columns={ 'Value': 'poultry_import' }, inplace=True)
df_import
```

### Chicken production

```python
df_production = load_file("fao_chicken_production_2017.csv")
```

```python
df_production.Value *= 1000
df_production.drop(columns=['Year', 'Unit'], inplace=True)
df_production.rename(columns={ 'Value': 'poultry_production' }, inplace=True)
df_production
```

### Protein supply

```python
df_protein = load_file("fao_protein_supply_2017.csv")
```

```python
df_protein.drop(columns=['Year'], inplace=True)
df_protein = df_protein.pivot_table(
    index='Area',
    columns='Item',
    values='Value',
    aggfunc='first'
).reset_index()

df_protein.rename(columns={ 
    'Animal Products': 'protein_animal_pct', 
    'Grand Total': 'protein_total_pct', 
    'Poultry Meat': 'protein_poultry_pct' 
}, inplace=True)

df_protein
```

### GDP

```python
df_gdp = load_file("wb_gdp_per_capita_2017.csv")
```

```python
df_gdp.dropna(subset=['2017'], inplace=True)
df_gdp.rename(columns={ 'Country Name': 'Area', '2017': 'GDP' }, inplace=True)
df_gdp.drop(columns=['Country Code'], inplace=True)
df_gdp
```

```python
df_gdp[df_gdp.Area.str.contains("China")]
```

### Stability

```python
df_stability = load_file("wb_stability_2017.csv")
```

```python
df_stability.dropna(subset=['2017 [YR2017]'], inplace=True)
df_stability.rename(columns={ 'Country Name': 'Area', '2017 [YR2017]': 'stability' }, inplace=True)
df_stability.drop(columns=['Country Code'], inplace=True)
df_stability.info()
```

### Urbanisation

```python
df_urban = load_file("wb_urbanization_2017.csv")
```

```python
df_urban.rename(columns={ 'REF_AREA_LABEL': 'Area', '2017': 'urban' }, inplace=True)
df_urban.drop(columns=['INDICATOR_LABEL'], inplace=True)
df_urban
```

## Merge

### Mapping

```python
# Apply mapping to World Bank tables BEFORE merging
wb_country_mapping = {
    'United States': 'United States of America',
    'United Kingdom': 'United Kingdom of Great Britain and Northern Ireland',
    'China': 'China, mainland',
    'Korea, Rep.': 'Republic of Korea',
    'Turkey': 'Türkiye',
    'Vietnam': 'Viet Nam',
    'Iran, Islamic Rep.': 'Iran (Islamic Republic of)',
    'Bolivia': 'Bolivia (Plurinational State of)',
    'Venezuela, RB': 'Venezuela (Bolivarian Republic of)',
    'Congo, Dem. Rep.': 'Democratic Republic of the Congo',
    'Congo, Rep.': 'Congo',
    "Cote d'Ivoire": "Côte d'Ivoire",
    'Tanzania': 'United Republic of Tanzania',
    'Moldova': 'Republic of Moldova',
    'Lao PDR': "Lao People's Democratic Republic",
    'Netherlands': 'Netherlands (Kingdom of the)',
    'Korea, Dem. Rep.': "Democratic People's Republic of Korea",
    'Egypt, Arab Rep.': 'Egypt',
    'Yemen, Rep.': 'Yemen',
    'Kyrgyz Republic': 'Kyrgyzstan',
    'Slovak Republic': 'Slovakia',
    'Bahamas, The': 'Bahamas',
    'Gambia, The': 'Gambia',
    'Turkiye': 'Türkiye',
    'Venezuela, RB': 'Venezuela (Bolivarian Republic of)',
    'Yemen, Rep.': 'Yemen',
}

# Apply to all World Bank dataframes
df_gdp['Area'] = df_gdp['Area'].replace(wb_country_mapping)
df_urban['Area'] = df_urban['Area'].replace(wb_country_mapping)
df_stability['Area'] = df_stability['Area'].replace(wb_country_mapping)
```

### Join

```python
df_master = df_population.copy()

df_master = df_master.merge(df_domestic, how='left', on='Area')
df_master = df_master.merge(df_dispo, how='left', on='Area')
df_master = df_master.merge(df_import, how='left', on='Area')
df_master = df_master.merge(df_production, how='left', on='Area')
df_master = df_master.merge(df_protein, how='left', on='Area')
df_master = df_master.merge(df_gdp, how='left', on='Area')
df_master = df_master.merge(df_stability, how='left', on='Area')
df_master = df_master.merge(df_urban, how='left', on='Area')
```

```python
df_master.dropna(inplace=True)
show_info(df_master)
```

### Check population coverage

```python
world_pop = 7_550_000_000  # 2017 world population

coverage = df_master['population'].sum() / world_pop * 100
print(f"Population coverage: {coverage:.1f}%")
```

## Feature Engineering 

```python
df_master.head()
```

### 1. Protein ratio (%)

```python
df_master['animal_protein_ratio'] = df_master['protein_animal_pct'] / df_master['protein_total_pct'] * 100
df_master['poultry_protein_ratio'] = df_master['protein_poultry_pct'] / df_master['protein_animal_pct'] * 100
```

### 2. Production per capita

```python
df_master['production_per_capita'] = df_master['poultry_production'] / df_master.population
```

### 3. Gap of consumption and production

```python
df_master['poultry_gap'] = df_master.poultry_supply_capita - df_master.production_per_capita
```

### 4. Import dependency

```python
df_master['import_dependency'] = df_master.poultry_import / df_master.poultry_domestic_supply * 100
```

### Clean up

```python
df_master.drop(columns=[
    'protein_animal_pct',
    'protein_total_pct', 
    'protein_poultry_pct',
    'poultry_domestic_supply'
], inplace=True)

df_master.rename(columns={ 'Area': 'country' }, inplace=True)
```
