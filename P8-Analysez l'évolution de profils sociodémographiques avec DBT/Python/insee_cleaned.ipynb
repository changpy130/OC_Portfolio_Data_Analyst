```python
# library
import pandas as pd
```

```python
# import file

df = pd.read_excel(
        'data_insee.xlsx',
        sheet_name="2022",
        header=None
    )
df.head()
```

```python
# get age groups

age_groups = df.iloc[4, 5:21].tolist()
print(age_groups)
print(len(age_groups))
```

```python
# get regions

regions = df.iloc[5:18, 0].tolist()
print(regions)
print(len(regions))

print(df.iloc[24, 0])
```

```python
# find start and end points(columns) of age groups for 2 genders

print("Male age groups:")
print(df.iloc[5, 26])  # start
print(df.iloc[5, 41])  # end = start + 15

print("Female age groups:")
print(df.iloc[5, 47])
print(df.iloc[5, 62])
```

```python
# retreive all data from 2022

all_data = []

regions_rows = range(5, 18)  # rows 5 to 17

for r_idx in regions_rows:
    region = df.iloc[r_idx, 0]
    row = df.iloc[r_idx]
    
    # Loop over age groups
    for ag_idx, age_group in enumerate(age_groups):
        pop_m = row.iloc[26 + ag_idx]
        pop_f = row.iloc[47 + ag_idx]
        
        # Add one row for M, one row for F
        all_data.append({'REGION': region, 'AGE_GROUP': age_group, 'GENDER': 'M', 'POPULATION': pop_m})
        all_data.append({'REGION': region, 'AGE_GROUP': age_group, 'GENDER': 'F', 'POPULATION': pop_f})

result = pd.DataFrame(all_data)
print(result.head(10))
```

```python
# retreive all data from 4 years

years = [2022, 2023, 2024, 2025]
all_data = []
regions_rows = range(5, 19)  # originally (5, 18), for adding DOM, put 19
age_groups = df.iloc[4, 5:21].tolist()
regions = df.iloc[5:18, 0].tolist().append(df.iloc[24, 0])  # plus DOM

for year in years:
    df = pd.read_excel(
        "data_insee.xlsx",
        sheet_name=str(year),
        header=None
    )
    
    for r_index in regions_rows:
        region = df.iloc[r_index, 0] if r_index < 18 else 'DOM'
        row = df.iloc[r_index] if r_index < 18 else df.iloc[24]

        for index, age in enumerate(age_groups):
            pop_m = row.iloc[26 + index]
            pop_f = row.iloc[47 + index]

            all_data.append(
                {
                    "YEAR": year,
                    "REGION": region,
                    "AGE_GROUP": age,
                    "GENDER": "M",
                    "POPULATION": pop_m
                }
            )
            all_data.append(
                {
                    "YEAR": year,
                    "REGION": region,
                    "AGE_GROUP": age,
                    "GENDER": "F",
                    "POPULATION": pop_f
                }
            )

result = pd.DataFrame(all_data)
result.head()
```

```python
# prepare age group mapping

age_oc = ['20-24 ans',
'25-29 ans',
'30-34 ans',
'35-39 ans',
'40-44 ans',
'45-49 ans',
'50-54 ans',
'55-59 ans',
'60 ans ou plus']

print(len(age_oc))
```

```python
# mapping age groups

age_mapping = dict()

for index, group in enumerate(age_groups):
    index_oc = index if index < 9 else 8
    age_mapping[group] = age_oc[index_oc]
```

```python
# apply into dataframe

result["AGE_GROUP"] = result["AGE_GROUP"].map(age_mapping)

# as_index means reset_index()
result = result.groupby(["YEAR", "REGION", "AGE_GROUP", "GENDER"], as_index=False)["POPULATION"].sum()
```

```python
# match region name

result['REGION'] = result['REGION'].replace('DOM', 'DROM')
```

```python
# Export csv

result.to_csv(
    "insee_population.csv",
    index=False,
    encoding='utf-8'
)
```
